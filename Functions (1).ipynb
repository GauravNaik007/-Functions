7. What are the advantages of using generators over regular functions?
  - Saves memory (doesn’t store all values at once)
  - Useful for large data or infinite loops


```
#Example
def count_up_to(n):
    num = 1
    while num <= n:
        yield num
        num += 1

for i in count_up_to(3):
    print(i)

```

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
  - map()
      - Used to apply a function to each item in an iterable (like list, tuple).
      - Returns a new iterable with the modified values.


```
#Example
nums = [1, 2, 3]
result = list(map(lambda x: x * 2, nums))
print(result)  # Output: [2, 4, 6]

```
  - filter()

    - Used to filter items in an iterable based on a condition (True/False).

    - Returns only those items that satisfy the condition.



```
# Example
nums = [1, 2, 3, 4]
result = list(filter(lambda x: x % 2 == 0, nums))
print(result)  # Output: [2, 4]

```
  - reduce()
    - Used to reduce the iterable to a single value by applying a function repeatedly.
    - Comes from the functools module.
  

```
#Example
from functools import reduce
nums = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, nums)
print(result)  # Output: 10

```

````python
#1.Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.
def sum_even_numbers(nums):
    return sum([x for x in nums if x % 2 == 0])
print(sum_even_numbers([1, 2, 3, 4, 5, 6]))
````

````python
#3. Implement a Python function that takes a list of integers and returns a new list containing the squares of each number.
def square_list(nums):
    return [x**2 for x in nums]

print(square_list([1, 2, 3]))  # Output: [1, 4, 9]
````

````python
#4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

print(is_prime(17))
````

````python
#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
class Fibonacci:
    def __init__(self, limit):
        self.a, self.b = 0, 1
        self.limit = limit
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.limit:
            raise StopIteration
        value = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return value

for num in Fibonacci(5):
    print(num)
````

````python
#6. Write a generator function in Python that yields the powers of 2 up to a given exponent.
def powers_of_two(n):
    for i in range(n+1):
        yield 2 ** i

for power in powers_of_two(4):
    print(power)
````

````python
#7. Implement a generator function that reads a file line by line and yields each line as a string.
def read_file_lines(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()
````

````python
#8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.
data = [(1, 4), (3, 1), (2, 9)]
sorted_data = sorted(data, key=lambda x: x[1])
print(sorted_data)
````

````python
#9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.
celsius = [0, 20, 37]
fahrenheit = list(map(lambda c: (9/5)*c + 32, celsius))
print(fahrenheit)
````

````python
#10. Create a Python program that uses `filter()` to remove all the vowels from a given string.
def remove_vowels(s):
    return ''.join(filter(lambda x: x.lower() not in 'aeiou', s))

print(remove_vowels("hello world"))
````
