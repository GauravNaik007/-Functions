##Question 1: What is Deep Learning? Briefly describe how it evolved and how it differs from traditional machine learning.

Ans:

- Deep Learning (DL) is a subset of Machine Learning (ML), which is a subset of Artificial Intelligence (AI). It uses artificial neural networks with multiple layers to learn complex patterns from data.

- Evolution of Deep Learning:

  - 1. Perceptron: Developed for basic classification.
  - 2. Multilayer neural networks: Introduced hidden layers to learn complex patterns.
  - 3. Backpropagation: Enabled networks to update weights and reduce errors.
  - 4. CNNs and RNNs: Used for image recognition and sequential data.
  - 5. Transformers: Introduced attention mechanisms and became important in NLP and Generative AI.

- Deep learning advanced due to increased computing power, large datasets and improved algorithms.

- Difference from traditional machine learning:

  - Traditional ML often requires manual feature extraction and works well with structured data. It can work with smaller datasets and generally requires less computing power.

  - Deep learning automatically learns features using multiple neural network layers. It is particularly effective for complex data such as images and text, but usually requires more data and computing power.

- Conclusion: Traditional ML is suitable for many structured-data problems, while deep learning is useful for learning complex patterns from large amounts of data.

##Question 2: Explain the basic architecture and functioning of a Perceptron. What are its limitations?

Answer:

- A perceptron is the simplest artificial neural network, consisting of a single neuron used for binary classification.
- Architecture: It contains input features, weights, bias, a summation unit and an activation function.
- Functioning:
  - 1. Takes input values and multiplies them by their weights.
  - 2. Adds the weighted inputs and bias:z = (w1 × x1) + (w2 × x2) + ... + (wn × xn) + b.
  - 3. Applies a step activation function to produce output 0 or 1.
  - 4. During training, adjusts weights and bias based on the prediction error.
- Limitations:
  - Can solve only linearly separable problems.
  - Cannot solve XOR using a single perceptron.
  - Cannot learn complex nonlinear patterns.
  - Has no hidden layer to learn intermediate features.

## Question 3: Describe the purpose of activation functions in neural networks. Compare Sigmoid, ReLU and Tanh.

Answer:

An activation function determines the output of a neuron after calculating the weighted sum and bias. It introduces nonlinearity, allowing neural networks to learn complex patterns.

1\. Sigmoid: Converts input into a value between 0 and 1. It is commonly used in binary classification output layers but can suffer from vanishing gradients.

Formula: `f(x) = 1 / (1 + e^(-x))`

2\. ReLU: Returns 0 for negative inputs and the input itself for positive inputs. It is commonly used in hidden layers and is computationally simple.

Formula: `f(x) = max(0, x)`

3\. Tanh: Converts input into a value between -1 and 1. It is zero-centred but can also suffer from vanishing gradients.

Formula: `f(x) = (e^x - e^(-x)) / (e^x + e^(-x))`

Conclusion: ReLU is commonly used in hidden layers, while Sigmoid is useful for binary classification outputs. Tanh is useful when outputs between -1 and 1 are needed.

Question 4: What is the difference between Loss function and Cost function in neural networks? Provide examples.

Answer:

- A loss function measures the error between the actual and predicted output for a single data point. A cost function measures the overall error across the training dataset, usually by averaging individual losses.
- Examples:
  - 1. Mean Squared Error (MSE): Used for regression problems. It calculates the average squared difference between actual and predicted values.
  - Formula (plain text): MSE = (1 / m) × [(y1 − ŷ1)² + (y2 − ŷ2)² + ... + (ym − ŷm)²]
  - 2. Binary Cross-Entropy: Used for binary classification. It measures the error between actual labels (0 or 1) and predicted probabilities.
- Conclusion: The loss or cost function helps the neural network measure its errors. During training, the optimizer updates weights and biases to minimize this error.

## Question 5: What is the role of optimizers in neural networks? Compare Gradient Descent, Adam, and RMSprop.

Answer:

Optimizers update the weights and biases of a neural network to minimize the loss function and improve model accuracy.

1\. Gradient Descent: Updates parameters using the gradient of the loss function. It can be slow for large datasets.

Formula (plain text): `New weight = Old weight − Learning rate × Gradient`

2\. RMSprop: Adjusts the learning rate for each parameter using a moving average of squared gradients. It helps in faster and more stable training.

3\. Adam: Combines Momentum and RMSprop. It adapts learning rates and generally provides efficient training.

Conclusion: Gradient Descent uses the gradient to update parameters, RMSprop adapts learning rates, and Adam combines momentum with adaptive learning rates.

```python
#Question 6: Write a Python program to implement a single-layer perceptron from scratch using NumPy to solve the logical AND gate.
import numpy as np

# AND gate input and output
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])

# Initialize weights and bias
weights = np.zeros(2)
bias = 0
learning_rate = 0.1
epochs = 20

# Step activation function
def activation(z):
    return 1 if z >= 0 else 0

# Train the perceptron
for epoch in range(epochs):
    for i in range(len(X)):
        z = np.dot(X[i], weights) + bias
        prediction = activation(z)

        error = y[i] - prediction

        weights += learning_rate * error * X[i]
        bias += learning_rate * error

# Test the trained perceptron
print("AND Gate Predictions:")

for inputs in X:
    z = np.dot(inputs, weights) + bias
    prediction = activation(z)

    print(inputs, "->", prediction)
```

```python
#Question 7: Implement and visualize Sigmoid, ReLU and Tanh activation functions using Matplotlib.
import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-5, 5, 200)

# Activation functions
sigmoid = 1 / (1 + np.exp(-x))
relu = np.maximum(0, x)
tanh = np.tanh(x)

# Plot the functions
plt.figure(figsize=(8, 5))

plt.plot(x, sigmoid, label="Sigmoid")
plt.plot(x, relu, label="ReLU")
plt.plot(x, tanh, label="Tanh")

plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)

plt.title("Activation Functions")
plt.xlabel("Input (x)")
plt.ylabel("Output f(x)")

plt.legend()
plt.grid(True)
plt.show()
```

```python
#Question 8: Use Keras to build and train a simple multilayer neural network on the MNIST digits dataset. Print the training accuracy.
import tensorflow as tf
from tensorflow.keras import layers, models

# Load the MNIST dataset
(x_train, y_train), (x_test, y_test) = (
    tf.keras.datasets.mnist.load_data()
)

# Normalize pixel values from 0-255 to 0-1
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Build a multilayer neural network
model = models.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

# Compile the model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train the model
model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=32,
    verbose=1
)

# Print training accuracy
loss, accuracy = model.evaluate(
    x_train, y_train, verbose=0
)

print("Training Accuracy:", accuracy)
```

```python
#Question 9: Visualize the loss and accuracy curves for a neural network model trained on the Fashion MNIST dataset. Interpret the training behavior.
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import layers, models

# Load Fashion MNIST
(x_train, y_train), (x_test, y_test) = (
    tf.keras.datasets.fashion_mnist.load_data()
)

# Normalize images
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Build the neural network
model = models.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train and store history
history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)

# Plot loss
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

# Plot accuracy
plt.subplot(1, 2, 2)
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("Accuracy Curve")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

"""Interpretation of training behavior
The training behavior can be interpreted by observing the loss and accuracy curves.
1. Learning: If training loss decreases and training accuracy increases, the model is learning patterns from the dataset.
2. Good generalization: If validation loss decreases and validation accuracy increases along with training performance, the model is learning patterns that also work on unseen validation data.
3. Overfitting: If training loss continues decreasing while validation loss starts increasing, the model may be memorizing the training data.
4. Underfitting: If both training and validation accuracy remain low, the model may not be learning enough useful patterns.
The plotted curves help determine whether the model is learning effectively and whether additional training is beneficial."""
```
