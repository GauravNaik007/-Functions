Question 1: What is a Decision Tree, and how does it work in the context of classification?

Ans:
  -  A Decision Tree is a supervised machine learning algorithm used for both classification and regression tasks. It works like a flowchart, with an inverted tree structure where each internal node represents a test on a feature (attribute), each branch represents the outcome of that test, and each leaf node represents a class label (in classification) or a continuous value (in regression).

How it works for classification:

1. The algorithm starts at the root node, which contains the entire dataset.
2. It selects the best feature to split the data on, using a criterion like Gini Impurity or Entropy/Information Gain.
3. The dataset is split into subsets based on the values of that feature.
4. This process repeats recursively for each subset (creating child nodes), choosing the best feature at each step.
5. The recursion stops when a stopping condition is met .for example, all data points in a node belong to one class, no more features remain, or a maximum depth is reached.
6. Each leaf node is then assigned the majority class of the samples that reach it.

  To classify a new or unseen data point, the algorithm starts from the root node and follows the appropriate branches based on the feature values of that data point. This continues until a leaf node is reached, and the class label of that leaf node becomes the prediction.

Question 2: Explain the concepts of Gini Impurity and Entropy as impurity measures.How do they impact the splits in a Decision Tree?

Ans:

- Both are impurity measures used to evaluate how "mixed" the classes are in a node. They help the decision tree decide which feature or threshold should be used to make a split.

Gini Impurity:

$$Gini = 1 - \sum_{i=1}^{n} p_i^2$$


- where $p_i$ is the probability of class $i$ in the node. It measures the probability of misclassifying a randomly chosen element if it were labeled randomly according to the class distribution in the node. Gini ranges from 0 (pure node) to 0.5 (maximum impurity for binary classification).

Entropy:

$$Entropy = -\sum_{i=1}^{n} p_i \log_2(p_i)$$

- It comes from information theory and measures the amount of disorder or uncertainty in the node. Entropy ranges from 0 (pure node) to 1 (maximum impurity for binary classification).

Impact on splits:

  - At each node, the algorithm evaluates all possible splits (feature + threshold combinations) and calculates the weighted impurity of the resulting child nodes.

  - The split that results in the lowest impurity (or highest reduction in impurity  see Information Gain below) is selected.

  - Gini is computationally slightly faster because it does not require a log calculation, and it is the default in scikit-learn's CART algorithm.Entropy is more sensitive to changes in class probability and is used in algorithms like ID3 and C4.5.

  - In practice, both usually produce very similar trees, so the choice between them does not always make a major difference in the final result.

Question 3: What is the difference between Pre-Pruning and Post-Pruning in Decision Trees? Give one practical advantage of using each.

Ans:

 Both are techniques used to prevent overfitting, where a tree becomes too complex and starts memorizing the training data instead of generalizing well to new data.

- Pre-Pruning (Early Stopping):

  - When applied: During tree construction.

  - How it works: Stops splitting a node early based on conditions like max_depth, min_samples_split, min_samples_leaf, or minimum impurity decrease.
  - Practical advantage: Saves training time and memory because the tree does not grow unnecessarily large.

- Post-Pruning (Backward Pruning):

  - When applied: After the full tree is built.
  - How it works: The full tree is built first, and then branches that do not improve performance are removed or collapsed.
  - Practical advantage: Can produce better generalization because the complete tree is considered before deciding which branches to remove.

-  key advantage each:

   - Pre-Pruning: Saves computation time and memory because a very large tree is avoided from the beginning. This can be useful for large datasets.

    - Post-Pruning: Can provide better generalization because the complete tree is first constructed, so useful splits are less likely to be removed too early.

Question 4: What is Information Gain in Decision Trees, and why is it important for choosing the best split?

Ans:

- Information Gain (IG) measures the reduction in entropy achieved by splitting the dataset on a particular feature.

$$IG = Entropy(parent) - \sum_{i=1}^{k} \frac{N_i}{N} \times Entropy(child_i)$$

- where $N$ is the total number of samples in the parent node, and $N_i$ is the number of samples in child node $i$.

Why it's important:

- It shows how much "information" or class purity is gained by splitting the data.
- At each node, the Decision Tree algorithm calculates the Information Gain for every possible feature or split and chooses the one with the highest IG.
- This helps move the data toward more homogeneous or pure subsets with each split, leading to a tree that can classify the data using fewer and more meaningful splits.
- Without Information Gain (or a similar criterion like Gini reduction), the algorithm would not have a proper way to decide which feature is the best for separating the classes.

Question 5: What are some common real-world applications of Decision Trees, and
what are their main advantages and limitations?

Ans:

Common Real-World Applications:


*   Finance: Credit scoring, loan approval, fraud detection
*   Healthcare: Disease diagnosis, patient risk prediction, treatment          recommendation
*   Marketing: Customer segmentation, churn prediction, targeted advertising
*   Retail/E-commerce: Recommendation systems, inventory management
*   Manufacturing: Quality control, predictive maintenance
*   HR: Employee attrition prediction, candidate screening

Advantages:

- Easy to understand and interpret — can be visualized as a flowchart, making it useful for explaining decisions to non-technical people
- Requires little data preprocessing — no need for feature scaling or normalization
- Can handle both numerical and categorical data
- Handles non-linear relationships well
- Naturally performs feature selection, since features that are not used for splitting are considered less useful for the tree

Limitations:

- Overfitting: Prone to creating overly complex trees that fit noise in the training data (this can be reduced using pruning)
- Instability: Small changes in the data can sometimes result in a very different tree structure (high variance)
- Bias toward features with more levels: Features with many categories can sometimes dominate the splits unfairly
- Not always optimal: The algorithm is greedy, meaning it chooses the locally best split at each step, which does not guarantee a globally optimal tree
- Poor performance on complex relationships alone: Decision Trees are often outperformed by ensemble methods like Random Forests or Gradient Boosting, which combine multiple trees to improve accuracy and stability

```python
"""Question 7: Write a Python program to:
● Load the Iris Dataset
● Train a Decision Tree Classifier with max_depth=3 and compare its accuracy to
a fully-grown tree.
(Include your Python code and output in the code box below.)"""

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

from sklearn.datasets import load_iris
data=load_iris()

df=pd.DataFrame(data.data,columns=data.feature_names)

X=df
y=data.target

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=1)

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Decision Tree with max_depth=3
classifier=DecisionTreeClassifier(criterion="gini",max_depth=3)
classifier.fit(X_train,y_train)

y_pred=classifier.predict(X_test)

accuracy_depth3=accuracy_score(y_test,y_pred)
print("Accuracy with max_depth=3:",accuracy_depth3)

# Fully-grown Decision Tree
classifier_full=DecisionTreeClassifier(criterion="gini")
classifier_full.fit(X_train,y_train)

y_pred_full=classifier_full.predict(X_test)

accuracy_full=accuracy_score(y_test,y_pred_full)
print("Accuracy of fully-grown tree:",accuracy_full)
```

Question 10: Imagine you’re working as a data scientist for a healthcare company that wants to predict whether a patient has a certain disease. You have a large dataset with mixed data types and some missing values.
Explain the step-by-step process you would follow to:
 * Handle the missing values
 * Encode the categorical features
 * Train a Decision Tree model
 *  Tune its hyperparameters
 * Evaluate its performance
 * And describe what business value this model could provide in the real-world
setting.

Ans:

Step 1: Understanding and Exploring the Data

- First, I would explore the dataset to understand what type of features are present.
-  For example, some features might be numerical, such as age, blood pressure, and glucose level, while others might be categorical, such as gender, blood group, and smoking status.
- I would also check the target variable, which would indicate whether the patient has the disease or not. I would check whether the classes are balanced or imbalanced and also check how many missing values are present in each column.

Step 2: Handling Missing Values

- For numerical features, I would fill the missing values using the mean or median. If there are outliers in the data, I would prefer using the median because it is less affected by extreme values.

- For categorical features, I would fill the missing values using the most frequent category (mode). Another option would be to create a separate category such as "Missing" or "Unknown".

- If a particular column has a very large number of missing values, I would consider removing that column instead of trying to fill most of its values.

Step 3: Encoding the Categorical Features

- Since the categorical values need to be converted into a form that the model can use, I would encode them into numerical values.

- For categories that do not have any particular order, such as gender or blood group, I would use One-Hot Encoding.
- For categories that have a natural order, such as disease severity (mild, moderate, severe), I would use Ordinal Encoding so that the order is maintained.

Step 4: Splitting the Data

- I would divide the dataset into training and testing sets, for example, using an 80-20 or 70-30 split.
-  Since healthcare datasets can have fewer disease-positive cases than disease-negative cases, I would use a stratified split so that both the training and testing sets have a similar proportion of the two classes.

Step 5: Training the Decision Tree Model

- After preprocessing the data, I would train a Decision Tree Classifier using the training data. I could use criteria such as Gini Impurity or Entropy to decide the best splits.

- Initially, I would train a basic Decision Tree to get a baseline result and understand how well the model performs before tuning it.

Step 6: Hyperparameter Tuning

- A Decision Tree can easily become too complex and overfit the training data. To reduce this problem, I would tune different hyperparameters such as:

    - max_depth — maximum depth of the tree
    - min_samples_split — minimum number of samples required to split a node
    - min_samples_leaf — minimum number of samples required in a leaf
    - criterion — Gini or Entropy
    - max_features — number of features considered when making a split

- I would use GridSearchCV or RandomizedSearchCV with cross-validation, such as 5-fold cross-validation, to find a good combination of these parameters.

Step 7: Evaluating the Model

- For a healthcare problem, accuracy alone may not be enough, especially when the classes are imbalanced. I would look at:

  -  Confusion Matrix — to see the true positives, false positives, true negatives, and false negatives.
  - Precision and Recall — Recall is especially important because missing a patient who actually has the disease (false negative) can be very risky.
  - F1-Score — to get a balance between precision and recall.
  - ROC-AUC Score — to measure how well the model can distinguish between the two classes.

- I would also compare the training and testing results to check whether the model is overfitting or underfitting.

Business Value in Real-World Setting

- This type of model could be useful for the healthcare company in several ways:

- It can help in the early detection of diseases, which can allow patients to receive treatment sooner.
- It can act as a first-level screening tool and identify patients who may need further testing.
- Since Decision Trees are relatively easy to understand, doctors and healthcare staff can see which features contributed to a particular prediction.
- It can help reduce unnecessary tests and allow healthcare resources to be focused more on high-risk patients.
- Overall, the model can support data-driven decision making, improve the efficiency of healthcare services, and potentially help with early intervention.
