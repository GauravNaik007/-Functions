# **Theory**




 Question 1: What is Ensemble Learning in machine learning? Explain the key idea
behind it.

Ans:
- Ensemble Learning is a machine learning technique in which multiple models are combined to make a final prediction instead of depending on only one model.

- The key idea behind ensemble learning is that multiple models working together can give better and more reliable results than a single model. Each model may make some mistakes, but by combining their predictions, the overall performance can be improved.

- For example, multiple Decision Trees can be trained on the data, and their predictions can be combined to produce the final prediction. In classification, the final result can be decided using majority voting, while in regression, the predictions can be averaged.

- Some common ensemble learning techniques are Bagging, Boosting, and Voting.

---
 Question 2: What is the difference between Bagging and Boosting?

Ans:

 - Bagging trains multiple models independently on different random samples of the data, mainly to reduce variance and overfitting, and the models can be trained in parallel. Examples include **Random Forest and Bagging Classifier**.

 - Boosting trains models **sequentially**, where each new model focuses more on the mistakes made by the previous models, mainly to reduce bias and improve accuracy. Examples include **AdaBoost, Gradient Boosting, and XGBoost**.
 ---
 Question 3: What is bootstrap sampling and what role does it play in Bagging methods
like Random Forest?

Ans:

- Bootstrap sampling is a technique in which multiple random samples are created from the original dataset with replacement, meaning the same data point can be selected more than once.
- In Bagging methods like Random Forest, each decision tree is trained on a different bootstrap sample, which creates diversity among the trees and helps reduce overfitting and improve the stability of predictions.
---
Question 4: What are Out-of-Bag (OOB) samples and how is OOB score used to evaluate ensemble models?

Ans:

- Out-of-Bag (OOB) samples are the data points that are  not selected in the bootstrap sample used to train a particular decision tree.
-  In Random Forest, these OOB samples are used to test that tree, and the OOB score measures how accurately the ensemble predicts these unused samples.
- Thus, OOB score provides an estimate of the model’s performance without needing a separate validation set.

---
Question 5: Compare feature importance analysis in a single Decision Tree vs. a
Random Forest.

Ans:

- In a single Decision Tree, feature importance is calculated based on how much each feature contributes to the splits in that one tree, so the importance can be affected by the tree’s structure.
- In a Random Forest, feature importance is calculated by combining the importance of a feature across many decision trees, making the result more stable and reliable.

```python
"""Question 6: Write a Python program to:
● Load the Breast Cancer dataset using
sklearn.datasets.load_breast_cancer()
● Train a Random Forest Classifier
● Print the top 5 most important features based on feature importance scores.
"""

import pandas as pd
from sklearn.datasets import load_breast_cancer
cancer_data=load_breast_cancer()

data=pd.DataFrame(cancer_data.data,columns=cancer_data.feature_names)
data["Target"]=cancer_data.target
X=data.drop("Target",axis=1)
y=data.Target

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=1)

from sklearn.ensemble  import RandomForestClassifier
clf=RandomForestClassifier()

clf.fit(X_train,y_train)

y_pred=clf.predict(X_test)

#get feature importance
feature_importance=pd.DataFrame({"Feature":X.columns,"Importance":clf.feature_importances_})


#sort features by importance
feature_importance=feature_importance.sort_values(by="Importance",ascending=-False)

print("Top 5 Most importance features: ")
print(feature_importance.head(5))
```

```python
"""Question 8: Write a Python program to:
● Train a Random Forest Classifier
● Tune hyperparameters max_depth and n_estimators using GridSearchCV
● Print the best parameters and final accuracy
"""
from sklearn.datasets import make_classification
X,y=make_classification(n_samples=1000,n_features=10,random_state=1)

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=1)

from sklearn.ensemble import  RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(X_train,y_train)
y_pred=rf.predict(X_test)


#hyperparameter tuning
from sklearn.model_selection import GridSearchCV
params={"max_depth":[1,2,3,4,10],
        "n_estimators":[10,20,30,50,60,100,200]}

grid_search=GridSearchCV(rf,param_grid=params,cv=5,verbose=3,scoring="accuracy")
grid_search.fit(X_train,y_train)

print("best params:",grid_search.best_params_)

final_model=grid_search.best_estimator_

y_pred_tuned=final_model.predict(X_test)
from sklearn.metrics import accuracy_score
print("final accuracy", accuracy_score(y_test,y_pred_tuned))
```

Question 10 :

You are working as a data scientist at a financial institution to predict loan
default. You have access to customer demographic and transaction history data.

You decide to use ensemble techniques to increase model performance.
Explain your step-by-step approach to:

- Choose between Bagging or Boosting
- Handle overfitting
- Select base models
- Evaluate performance using cross-validation
- Justify how ensemble learning improves decision-making in this real-world context.

Ans:


As a data scientist, I would follow these steps to build an ensemble model for predicting loan default.

**1. Choose between Bagging and Boosting**

First, I would understand the nature of the data and the problem. Since loan default is a classification problem, both Bagging and Boosting can be used. Bagging is useful when I want to reduce variance and avoid depending on a single model. Boosting is useful when I want to improve the model step by step by focusing on incorrectly predicted customers. I would try both methods and select the one that gives better validation performance.

**2. Handle overfitting**

Overfitting can happen when the model learns the training data too closely. I would control it by limiting the depth of decision trees, choosing a suitable number of estimators and using cross-validation. In boosting, I would also control parameters such as learning rate and tree depth. I would make sure that the model performs well on unseen data, not only on training data.

**3. Select base models**

For Bagging, I would mainly consider Decision Trees as base models because they can capture different patterns in customer and transaction data. For Boosting, Decision Trees are also commonly used as weak learners. Depending on the data, I could compare different models and select the one that gives good performance without too much complexity.

**4. Evaluate using cross-validation**

I would divide the training data into multiple folds and use cross-validation. For example, with 5-fold cross-validation, the model is trained on four parts and tested on the remaining part. This process is repeated five times. I would compare metrics such as accuracy, precision, recall, F1-score and ROC-AUC. For loan default prediction, recall and precision are important because incorrectly identifying a risky or safe customer can have financial consequences.

**5. How ensemble learning improves decision-making**

Ensemble learning combines predictions from multiple models instead of depending on only one model. This can make predictions more reliable and reduce the effect of individual model errors. In a financial institution, this can help identify customers who are more likely to default and support better loan approval decisions. It can also help the institution reduce financial losses by identifying high-risk customers earlier.

Overall, I would train and compare Bagging and Boosting models, control overfitting, evaluate them using cross-validation and choose the model that gives the best balance between performance and reliability.
