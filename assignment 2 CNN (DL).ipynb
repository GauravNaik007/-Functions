# **THEORY**

## Q1. Explain the role of filters and feature maps in a CNN.

Ans:

- Filters (Kernels): Filters are small, learnable matrices that slide over an image and perform convolution. They detect features such as edges, lines, corners, and textures.

- Feature Maps: A feature map is the output produced when a filter is applied to an image. It shows where the detected feature appears. Each filter produces its own feature map.

- Working: The filter slides across the image, performs calculations, and generates a feature map. Multiple filters help a CNN learn different image features.

- Example: A \\(5 \times 5\\) image convolved with a \\(3 \times 3\\) filter, using stride 1 and no padding, produces a \\(3 \times 3\\) feature map.

---
## Q2. Explain the concepts of padding and stride in CNNs. How do they affect the output dimensions of feature maps?

Ans:

Padding means adding extra pixels, usually zeros, around the input image. It helps control the output size and preserve spatial dimensions.
- Valid padding: No extra pixels are added, so the feature map becomes smaller.
- Same padding: Padding is added to keep the output size approximately the same as the input when stride is 1.
Stride is the number of pixels the filter moves at each step.
- Stride 1 moves the filter one pixel at a time and usually gives a larger output.
- Stride 2 moves it two pixels at a time and gives a smaller output.
Output dimension:
O = floor((I + 2P - K) / S) + 1

- Where:
  - O = output size
  - I = input size
  - P = padding
  - K = filter (kernel) size
  - S = stride
- The floor means round down to the nearest whole number.
- The output is (3 X 3).

Q3. Define receptive field in the context of CNNs. Why is it important for deep architectures?

Ans:

- The receptive field is the part of the input image that affects a particular neuron in a CNN.
- In early layers, a neuron usually sees a small area of the image.
- As more convolutional layers are stacked, the receptive field becomes larger.
- Importance:
  - Small receptive fields help detect local features such as edges and textures.
  - Larger receptive fields help detect complex patterns and whole objects.
  - They allow deeper CNNs to understand both small details and broader image context.

Q4. Discuss how filter size and stride influence the number of parameters in a CNN.

Ans:

- The number of trainable parameters in a convolutional layer depends mainly on the filter size, input channels, number of filters, and biases.
Parameters = (Kh × Kw × Cin + 1) × Cout

- Where:
  - Kh = filter height
  - Kw = filter width
  - Cin = number of input channels
  - Cout = number of filters
  - +1 = one bias per filter

- Filter size: Larger filters contain more weights, increasing the number of parameters. For example, a (5X5) filter has more weights than a (3X3) filter.
- Stride: Changing stride affects the size of the output feature map and computation. It does not change the number of weights in the filter itself.
Therefore, filter size affects parameter count, while stride mainly affects output dimensions and computational cost.

## Q5. Compare and contrast different CNN-based architectures like LeNet, AlexNet, and VGG in terms of depth, filter sizes, and performance.

Ans:

- LeNet: LeNet is a relatively shallow CNN designed mainly for handwritten digit recognition. It uses convolution and pooling layers to extract simple image features. It is lightweight and requires less computation.

- AlexNet: AlexNet is deeper than LeNet and was developed for large-scale image classification. It uses convolutional filters of different sizes, ReLU activation, and dropout. It achieved better performance on complex image-classification tasks.

- VGG: VGG is a deeper CNN that mainly uses small (3X3) filters stacked across many layers. This helps it learn complex image features. It performs well on image classification but requires more parameters and computational resources.

-  LeNet is simple and suitable for basic image tasks, AlexNet introduced a deeper and more powerful architecture, and VGG uses many small filters in a deeper network to learn detailed features.

````python
#Q6. Using Keras, build and train a simple CNN model on the MNIST dataset from scratch. Include code for module creation, compilation, training, and evaluation.
import tensorflow as tf
from tensorflow.keras import layers, models

# Load MNIST
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize and add channel dimension
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
x_train = x_train[..., None]
x_test = x_test[..., None]

# Create CNN
model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])

# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train
model.fit(x_train, y_train, epochs=5, batch_size=64,
          validation_split=0.1)

# Evaluate
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print("Test accuracy:", test_accuracy)
````

````python
#Q7. Load and preprocess the CIFAR-10 dataset using Keras, and create a CNN model to classify RGB images. Show your preprocessing and architecture.
import tensorflow as tf
from tensorflow.keras import layers, models

# Load dataset
(x_train, y_train), (x_test, y_test) = \
    tf.keras.datasets.cifar10.load_data()

# Normalize pixel values
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Build CNN
model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),
    layers.Conv2D(32, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation="relu", padding="same"),
    layers.MaxPooling2D(),
    layers.Conv2D(128, 3, activation="relu", padding="same"),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(x_train, y_train, epochs=10, batch_size=64,
          validation_split=0.1)

model.evaluate(x_test, y_test)
````

````python
#Q8. Using PyTorch, write a script to define and train a CNN on the MNIST dataset. Include model definition, data loaders, training loop, and accuracy evaluation.
import torch
from torch import nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms



# Prepare data
transform = transforms.ToTensor()

train_data = datasets.MNIST(
    root="data", train=True, download=True, transform=transform
)
test_data = datasets.MNIST(
    root="data", train=False, download=True, transform=transform
)

train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64)

# Define CNN
class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(64 * 7 * 7, 10)
        )

    def forward(self, x):
        return self.net(x)

model = CNN()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
criterion = nn.CrossEntropyLoss()

# Train
for epoch in range(5):
    model.train()
    for images, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

# Evaluate accuracy
model.eval()
correct = total = 0

with torch.no_grad():
    for images, labels in test_loader:
        predictions = model(images).argmax(dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

print("Test accuracy:", correct / total)
````

````python
""""Question 9: Given a custom image dataset stored in a local directory, write code using
Keras ImageDataGenerator to preprocess and train a CNN model""""

from google.colab import files
import zipfile
import os
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. Upload the ZIP file containing dataset/cats and dataset/dogs
uploaded = files.upload()
zip_path = next(iter(uploaded))

# 2. Extract the dataset
with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall("/content")

# 3. Set the dataset path
dataset_path = "/content/dataset"

# Check that the class folders exist
print("Dataset folders:", os.listdir(dataset_path))

# 4. Create ImageDataGenerator with rescaling and validation split
datagen = ImageDataGenerator(
    rescale=1.0 / 255,
    validation_split=0.2
)

# 5. Load training images
train_data = datagen.flow_from_directory(
    dataset_path,
    target_size=(150, 150),
    batch_size=32,
    class_mode="binary",
    subset="training"
)

# 6. Load validation images
validation_data = datagen.flow_from_directory(
    dataset_path,
    target_size=(150, 150),
    batch_size=32,
    class_mode="binary",
    subset="validation"
)

# 7. Display one batch of training images
images, labels = next(train_data)

plt.figure(figsize=(10, 6))

for i in range(min(8, len(images))):
    plt.subplot(2, 4, i + 1)
    plt.imshow(images[i])
    plt.title("Dog" if labels[i] == 1 else "Cat")
    plt.axis("off")

plt.tight_layout()
plt.show()

# 8. Print dataset information
print("Class labels:", train_data.class_indices)
print("Training images:", train_data.samples)
print("Validation images:", validation_data.samples)
````
