## Q1. What is a Vector Database (VectorDB) and how is it different from traditional databases?

### 1. Vector Database

A Vector Database (VectorDB) is a type of database used to store and search numerical representations of data called *vector embeddings*. These embeddings represent the meaning or features of text, images, audio, and other data.

Vector databases are commonly used in AI and Machine Learning for similarity-based searches.

### 2. Working of a Vector Database

1. **Data Input:** Text, images, or documents are provided.
2. **Embedding Generation:** An embedding model converts the data into numerical vectors.
3. **Storage:** The generated vectors are stored in the database.
4. **Similarity Search:** The database compares vectors using measures such as cosine similarity or Euclidean distance.
5. **Result Retrieval:** The most similar results are returned to the user.

### 3. Difference Between VectorDB and Traditional Databases

| Traditional Database | Vector Database |
|---|---|
| Stores data mainly in tables. | Stores numerical vector embeddings. |
| Uses SQL queries and exact matching. | Uses similarity-based searching. |
| Mainly suitable for structured data. | Useful for text, images, and other data represented as vectors. |
| Retrieves records using specified conditions. | Retrieves data based on similarity in meaning or features. |
| Examples: MySQL, PostgreSQL. | Examples: ChromaDB, Pinecone, Weaviate. |

### Conclusion

A Vector Database stores and retrieves data using vector embeddings and similarity search. Unlike traditional databases, which mainly use structured queries and specified conditions, VectorDBs help find information based on semantic or feature similarity.

## Q2. Explain the Various Types of VectorDBs Available and Describe Their Suitability for Different Use Cases.

Vector databases can be classified based on their architecture, storage methods, and search capabilities. Different VectorDBs are suitable for different AI and Machine Learning applications.

### 1. ChromaDB
ChromaDB is an open-source vector database that is simple to use and integrates well with Python and AI frameworks.

**Use cases:** Small to medium-sized AI projects, semantic search, document retrieval, and RAG-based chatbots.

### 2. Pinecone
Pinecone is a managed vector database designed for scalable and efficient similarity search. It reduces the need to manage database infrastructure manually.

**Use cases:** Production AI applications, large-scale recommendation systems, and enterprise search.

### 3. Weaviate
Weaviate is an open-source vector database that supports vector search, keyword search, and metadata filtering.

**Use cases:** Knowledge search systems, AI assistants, and applications that combine semantic and keyword-based retrieval.

### 4. Milvus
Milvus is an open-source vector database designed to handle large collections of high-dimensional vectors efficiently.

**Use cases:** Large-scale image search, recommendation engines, and applications that require high-performance vector retrieval.

### 5. FAISS
FAISS is a library developed by Meta for efficient similarity search and clustering of dense vectors. It is primarily a vector search library rather than a complete database system.

**Use cases:** Research projects, machine learning experiments, and high-speed similarity search on large vector collections.

### Conclusion
Different VectorDBs serve different requirements. ChromaDB is suitable for beginner and smaller AI projects, Pinecone for managed production applications, Weaviate for combined search methods, Milvus for large-scale vector search, and FAISS for efficient similarity search and research.

````python
#Q7. Demonstrate How to Download and Fine-Tune a Hugging Face Model for a Text Generation Task
"""Hugging Face provides pre-trained models that can be downloaded and fine-tuned on a custom dataset. Fine-tuning helps a model adapt to a specific type of text generation task.
Python Code
First, install the required libraries:"""

!pip install transformers datasets torch
"""The following code downloads the GPT-2 model, fine-tunes it on sample text, and generates new text.
Output
The training progress and generated text will vary depending on the environment and model execution."""

from transformers import (
    GPT2Tokenizer,
    GPT2LMHeadModel,
    Trainer,
    TrainingArguments,
    DataCollatorForLanguageModeling
)
from datasets import Dataset

# Download the pre-trained model
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

tokenizer.pad_token = tokenizer.eos_token
model.config.pad_token_id = tokenizer.eos_token_id

# Create a sample dataset
texts = [
    "Artificial intelligence helps computers learn.",
    "Machine learning is a branch of artificial intelligence.",
    "Python is widely used for machine learning.",
    "Text generation is an application of language models."
]

dataset = Dataset.from_dict({"text": texts})

# Tokenize the dataset
def tokenize_data(examples):
    return tokenizer(
        examples["text"],
        truncation=True,
        padding="max_length",
        max_length=32
    )

tokenized_dataset = dataset.map(
    tokenize_data, batched=True
)

# Configure fine-tuning
training_args = TrainingArguments(
    output_dir="./gpt2_finetuned",
    num_train_epochs=1,
    per_device_train_batch_size=2,
    report_to="none"
)

data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

# Fine-tune the model
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
    data_collator=data_collator
)

trainer.train()

# Generate text
prompt = "Artificial intelligence"
inputs = tokenizer(prompt, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=20,
    do_sample=False,
    pad_token_id=tokenizer.eos_token_id
)

print(tokenizer.decode(
    outputs[0], skip_special_tokens=True
))
````

````python
#Q8. Create a Custom LLM Using Ollama and Llama 2
"Ollama allows users to run Large Language Models locally. In this experiment, Llama 2 is downloaded, customized with a Modelfile, and used to generate text from a prompt."
# Install Ollama

# Install zstd and Ollama dependencies
!apt-get update -qq
!apt-get install -y -qq zstd
!curl -fsSL https://ollama.com/install.sh | sh
!pip install -q chromadb ollama

import subprocess
import time
import chromadb
import ollama

# Start Ollama server
server = subprocess.Popen(
    ["ollama", "serve"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

# Wait for the server to start
time.sleep(5)

# Download Llama 3
subprocess.run(["ollama", "pull", "llama3"], check=True)

# Create ChromaDB collection
client = chromadb.Client()
collection = client.create_collection(
    name="rag_documents"
)

# Insert sample documents
documents = [
    "Machine learning allows computers to learn from data.",
    "Python is a popular programming language.",
    "RAG retrieves relevant documents before generating an answer."
]

collection.add(
    ids=["1", "2", "3"],
    documents=documents
)

# Retrieve relevant documents
query = "What is Retrieval-Augmented Generation?"

results = collection.query(
    query_texts=[query],
    n_results=2
)

context = "\n".join(results["documents"][0])

# Generate an answer using Llama 3
prompt = f"""
Answer the question using only the context below.

Context:
{context}

Question: {query}
Answer:
"""

response = ollama.generate(
    model="llama3",
    prompt=prompt
)

print("Retrieved Documents:")
print(context)

print("\nGenerated Answer:")
print(response["response"])
````

````python
#Q9. Implement a Basic RAG (Retrieval-Augmented Generation) System Using Ollama with Llama 3
"""Retrieval-Augmented Generation (RAG) is a technique that combines information retrieval with text generation. It retrieves relevant information from a collection of documents and provides that information to a Large Language Model (LLM) to generate a context-based answer.

In this experiment, Ollama with Llama 3 is used for text generation, and ChromaDB is used to retrieve relevant documents."""

# Install required libraries
!curl -fsSL https://ollama.com/install.sh | sh
!pip install chromadb ollama

import subprocess
import time
import chromadb
import ollama

# Start Ollama server
server = subprocess.Popen(
    ["ollama", "serve"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

time.sleep(5)

# Download Llama 3
subprocess.run(["ollama", "pull", "llama3"], check=True)

# Create ChromaDB collection
client = chromadb.Client()
collection = client.create_collection(name="rag_documents")

# Add sample documents
documents = [
    "Machine learning allows computers to learn from data.",
    "Python is a popular programming language.",
    "RAG retrieves relevant documents before generating an answer."
]

collection.add(
    ids=["1", "2", "3"],
    documents=documents
)

# Retrieve relevant documents
query = "What is Retrieval-Augmented Generation?"

results = collection.query(
    query_texts=[query],
    n_results=2
)

context = "\n".join(results["documents"][0])

# Generate answer using Llama 3
prompt = f"""
Answer the question using the following context.

Context:
{context}

Question: {query}
Answer:
"""

response = ollama.generate(
    model="llama3",
    prompt=prompt
)

print("Retrieved Documents:")
print(context)

print("\nGenerated Answer:")
print(response["response"])
````

````python
#Question 10: A health-tech startup wants to build a chatbot that can answer user queries based on medical research articles. Propose and explain a solution using Hugging Face models for understanding, VectorDB for retrieval, and Ollama for generation.

"""A health-tech startup can build a medical research chatbot using Hugging Face for understanding text, ChromaDB for retrieving relevant research articles, and Ollama with Llama 3 for generating answers. This system uses Retrieval-Augmented Generation (RAG) to provide answers based on medical research documents.
1. Proposed Solution
1. Hugging Face — Text Understanding: Use a pre-trained embedding model such as all-MiniLM-L6-v2 to convert medical research articles and user queries into numerical vectors.
2. VectorDB (ChromaDB) — Information Retrieval: Store the document embeddings and retrieve relevant medical research passages based on the user's question.
3. Ollama with Llama 3 — Answer Generation: Pass the retrieved passages and user's question to Llama 3, which generates a context-based response.
4. User Interface: Display the generated answer along with the relevant article references so users can check the original research."""

# Install only the additional libraries needed for Q10
!pip install -q chromadb sentence-transformers ollama

import chromadb
import ollama
from sentence_transformers import SentenceTransformer

# Load the Hugging Face embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Sample medical research articles
articles = [
    "Regular physical activity can improve cardiovascular health.",
    "Adequate sleep supports immune function and general health.",
    "Blood glucose monitoring is important in diabetes management."
]

# Create the vector database
client = chromadb.Client()
collection = client.create_collection(
    name="medical_research"
)

# Convert articles into embeddings
embeddings = model.encode(articles).tolist()

# Store articles in ChromaDB
collection.add(
    ids=["article1", "article2", "article3"],
    documents=articles,
    embeddings=embeddings
)

# Ask a question and retrieve relevant information
query = "How does exercise affect heart health?"
query_embedding = model.encode([query]).tolist()

results = collection.query(
    query_embeddings=query_embedding,
    n_results=2
)

context = "\n".join(results["documents"][0])

# Generate an answer using the existing Ollama model
prompt = f"""
Answer using the research context below.
If the context is insufficient, say so.

Context:
{context}

Question: {query}
Answer:
"""

response = ollama.generate(
    model="llama3",
    prompt=prompt
)

print("Retrieved Research:")
print(context)
print("\nChatbot Answer:")
print(response["response"])
````
