### Question 1: What is a CNN, and how does it differ from traditional fully connected neural networks in architecture and performance on image data?

**Answer:**

- A Convolutional Neural Network (CNN) is a deep learning model mainly used for image classification. It automatically extracts features such as edges, shapes, and textures from images to identify objects.

- **Architecture of CNN:**

  - A CNN consists of convolutional layers that extract image features using filters, ReLU activation layers that introduce non-linearity, pooling layers that reduce feature-map size, and fully connected layers that perform classification. The output layer predicts the image class.

- **Difference from fully connected neural networks:**

  - Traditional fully connected networks connect every neuron to all neurons in the previous layer and usually flatten the image into a vector. CNNs use convolutional filters that share weights and preserve spatial information while extracting features.

  - CNNs generally require fewer parameters and are more efficient for image processing. They learn local patterns such as edges and shapes, making them better suited for image classification.

### Question 2: Discuss the architecture of LeNet-5 and explain how it laid the foundation for modern deep learning models in computer vision. Include references to its original research paper.

**Answer:**

LeNet-5 is an early Convolutional Neural Network (CNN) developed by Yann LeCun and his collaborators in 1998. It was designed mainly for handwritten digit recognition, such as identifying digits from 0 to 9.

**Architecture of LeNet-5:**

LeNet-5 takes a \(32 \times 32\) grayscale image as input and processes it through the following layers:

1. **C1 – Convolutional layer:** Uses 6 filters of size \(5 \times 5\) to extract basic features such as edges and strokes.
2. **S2 – Subsampling layer:** Uses average pooling to reduce the feature-map dimensions.
3. **C3 – Convolutional layer:** Uses 16 filters to learn more complex features.
4. **S4 – Subsampling layer:** Further reduces the spatial dimensions.
5. **C5 – Convolutional layer:** Produces 120 feature units.
6. **F6 – Fully connected layer:** Contains 84 units and combines the extracted features.
7. **Output layer:** Classifies the image into one of the ten digit categories.

**How LeNet-5 laid the foundation for modern CNNs:**

LeNet-5 introduced important ideas that are still used in modern computer vision models. Convolutional layers automatically learn image features, while shared filter weights reduce the number of parameters. Pooling reduces image dimensions, and successive layers learn increasingly complex patterns.

It also demonstrated that a neural network could learn features directly from images and perform classification through end-to-end training. These principles influenced later CNN architectures such as AlexNet, VGGNet, and GoogLeNet.

**Original research paper:**

LeCun, Y., Bottou, L., Bengio, Y., and Haffner, P. (1998). *Gradient-Based Learning Applied to Document Recognition*. Proceedings of the IEEE, 86(11), 2278–2324. DOI: 10.1109/5.726791.

**Conclusion:** LeNet-5 established the basic architecture of CNNs by combining convolution, subsampling, and classification. It became an important foundation for the development of deeper neural networks used in modern computer vision.

### Q3. Compare AlexNet and VGGNet

Ans:

- **AlexNet** has 5 convolutional layers and 3 fully connected layers, with approximately 60 million parameters. It introduced important techniques such as ReLU activation, dropout, data augmentation, and GPU-based training. These helped it achieve strong performance on ImageNet and popularize deep CNNs. However, its large fully connected layers required substantial memory and computation.

- **VGGNet** uses a deeper and more uniform architecture. VGG16 contains 13 convolutional layers and 3 fully connected layers, with approximately 138 million parameters. It mainly uses small 3×3 convolution filters stacked together, allowing it to learn complex image features. VGGNet achieved better image-recognition performance than AlexNet in many ImageNet evaluations, but its greater depth and parameter count made it more computationally expensive.

- **In summary:** AlexNet was an important early breakthrough, while VGGNet showed that increasing network depth using small filters could improve image classification.

### Q4. Transfer Learning in Image Classification

Ans:

**Transfer learning** is a technique where a model trained on a large dataset is reused for a new image-classification task. Instead of training a CNN from scratch, we use a pretrained model such as VGG16 or ResNet, which has already learned useful features like edges, shapes, and textures.

The pretrained model is adapted to the new task by replacing its final classification layer with one suited to the new classes. Its earlier layers can be kept frozen, or some layers can be fine-tuned using the new dataset.

**How it reduces computational cost:**
- The model does not need to learn all image features from the beginning.
- Freezing pretrained layers reduces the amount of training required.
- Training usually takes less time and computing power.

**How it improves performance with limited data:**
- Pretrained models already contain useful visual features learned from many images.
- The new model can learn the target classes from fewer examples.
- It can generalize better than a model trained from scratch, which may overfit when data is limited.

Thus, transfer learning makes image classification faster and often more accurate, especially when the new dataset is small.

### Q5. Residual Connections in ResNet

**Residual connections** are shortcut connections in ResNet that allow the input of a layer to bypass one or more convolutional layers and be added directly to their output.

Instead of learning the complete mapping \(H(x)\), the layers learn a **residual function**:
\\[ \boxed{H(x)=F(x)+x} \\]

Here, \(x\) is the original input and \(F(x)\) is the output of the convolutional layers. The shortcut adds \(x\) to \(F(x)\).

**How they help with the vanishing gradient problem:**

- In very deep CNNs, gradients can become extremely small as they travel backward through many layers. This makes early layers difficult to train.
- The shortcut provides a direct path for the gradient to flow backward, helping it reach earlier layers.
- This makes it easier to train very deep networks and reduces the degradation problem, where adding more layers does not improve performance.

Thus, residual connections help ResNet train deeper CNNs more effectively by allowing information and gradients to pass through shortcut paths.

```python
###Question 6: Implement the LeNet-5 architectures using Tensorflow or PyTorch to classify the MNIST dataset. Report the accuracy and training time.
import tensorflow as tf
import time

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize pixel values to 0–1
x_train = x_train / 255.0
x_test = x_test / 255.0

# Add channel dimension and pad 28x28 images to 32x32
x_train = x_train[..., tf.newaxis]
x_test = x_test[..., tf.newaxis]

x_train = tf.pad(x_train, [[0, 0], [2, 2], [2, 2], [0, 0]])
x_test = tf.pad(x_test, [[0, 0], [2, 2], [2, 2], [0, 0]])

# Build LeNet-5 model
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(32, 32, 1)),

    tf.keras.layers.Conv2D(6, kernel_size=5, activation="tanh"),
    tf.keras.layers.AveragePooling2D(pool_size=2),

    tf.keras.layers.Conv2D(16, kernel_size=5, activation="tanh"),
    tf.keras.layers.AveragePooling2D(pool_size=2),

    tf.keras.layers.Conv2D(120, kernel_size=5, activation="tanh"),
    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(84, activation="tanh"),
    tf.keras.layers.Dense(10, activation="softmax")
])

# Compile model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train and measure training time
start_time = time.time()

model.fit(
    x_train, y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)

training_time = time.time() - start_time

# Evaluate on test data
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)

print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
print(f"Training Time: {training_time:.2f} seconds")
```

```python
#Question 7: Use a pre-trained VGG16 model (via transfer learning) on a small custom dataset (e.g., flowers or animals). Replace the top layers and fine-tune the model.Include your code and result discussion.

import tensorflow as tf
import time
import os

# Download and extract the flower dataset
url = "https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz"
data_dir = tf.keras.utils.get_file(
    "flower_photos", origin=url, untar=True
)

# Load images and split into training and validation sets
img_size = (224, 224)
batch_size = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=img_size,
    batch_size=batch_size
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=img_size,
    batch_size=batch_size
)

class_names = train_ds.class_names
num_classes = len(class_names)

train_ds = train_ds.prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.prefetch(tf.data.AUTOTUNE)

# Load VGG16 pretrained on ImageNet, without its original classifier
base_model = tf.keras.applications.VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# Initially freeze pretrained convolutional layers
base_model.trainable = False

# Replace the top classifier for our flower classes
inputs = tf.keras.Input(shape=(224, 224, 3))

x = tf.keras.layers.RandomFlip("horizontal")(inputs)
x = tf.keras.layers.RandomRotation(0.1)(x)
x = tf.keras.applications.vgg16.preprocess_input(x)

x = base_model(x, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dense(128, activation="relu")(x)
x = tf.keras.layers.Dropout(0.3)(x)
outputs = tf.keras.layers.Dense(
    num_classes, activation="softmax"
)(x)

model = tf.keras.Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train the new classifier while VGG16 layers are frozen
start_time = time.time()

model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)

# Fine-tune the last VGG16 convolutional block
base_model.trainable = True

for layer in base_model.layers:
    if not layer.name.startswith("block5"):
        layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)

training_time = time.time() - start_time

# Evaluate on the validation set
val_loss, val_accuracy = model.evaluate(val_ds)

print("Flower classes:", class_names)
print(f"Validation accuracy: {val_accuracy * 100:.2f}%")
print(f"Total training time: {training_time:.2f} seconds")

""""""
```

```python
#Question 8: Write a program to visualize the filters and feature maps of the first convolutional layer of AlexNet on an example input image.


import torch
import torchvision
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files

# Upload an example image
uploaded = files.upload()
image_path = next(iter(uploaded))

# Load pretrained AlexNet
weights = torchvision.models.AlexNet_Weights.DEFAULT
model = torchvision.models.alexnet(weights=weights)
model.eval()

# Prepare the input image
image = Image.open(image_path).convert("RGB")
transform = weights.transforms()
input_tensor = transform(image).unsqueeze(0)

# Get the first convolutional layer
first_conv = model.features[0]

# Extract its filters
filters = first_conv.weight.detach().cpu()

# Display 16 filters
fig, axes = plt.subplots(4, 4, figsize=(8, 8))

for i, ax in enumerate(axes.flat):
    # Convert filter from (channels, height, width) to image format
    filt = filters[i].permute(1, 2, 0)

    # Normalize values for display
    filt = (filt - filt.min()) / (filt.max() - filt.min() + 1e-8)

    ax.imshow(filt)
    ax.set_title(f"Filter {i + 1}")
    ax.axis("off")

plt.suptitle("First Convolutional Layer: AlexNet Filters")
plt.tight_layout()
plt.show()

# Obtain feature maps from the first convolutional layer
with torch.no_grad():
    feature_maps = first_conv(input_tensor)

feature_maps = feature_maps[0].cpu()

# Display 16 feature maps
fig, axes = plt.subplots(4, 4, figsize=(10, 10))

for i, ax in enumerate(axes.flat):
    ax.imshow(feature_maps[i], cmap="gray")
    ax.set_title(f"Feature Map {i + 1}")
    ax.axis("off")

plt.suptitle("Feature Maps for the Example Image")
plt.tight_layout()
plt.show()
```

```python
#Question 9: Train a GoogLeNet (Inception v1) or its variant using a standard dataset like CIFAR-10. Plot the training and validation accuracy over epochs and analyze overfitting or underfitting.
import torch
import torch.nn as nn
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import time

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load CIFAR-10 and normalize images
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

train_data = torchvision.datasets.CIFAR10(
    root="./data", train=True, download=True, transform=transform
)
val_data = torchvision.datasets.CIFAR10(
    root="./data", train=False, download=True, transform=transform
)

train_loader = torch.utils.data.DataLoader(
    train_data, batch_size=64, shuffle=True
)
val_loader = torch.utils.data.DataLoader(
    val_data, batch_size=64, shuffle=False
)

# GoogLeNet (Inception v1), trained from scratch
model = torchvision.models.googlenet(
    weights=None, aux_logits=False, num_classes=10
).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

train_accs = []
val_accs = []
epochs = 5

start_time = time.time()

for epoch in range(epochs):
    # Training
    model.train()
    correct = 0
    total = 0

    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    train_accuracy = 100 * correct / total
    train_accs.append(train_accuracy)

    # Validation
    model.eval()
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)

            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    val_accuracy = 100 * correct / total
    val_accs.append(val_accuracy)

    print(
        f"Epoch {epoch + 1}/{epochs} | "
        f"Train accuracy: {train_accuracy:.2f}% | "
        f"Validation accuracy: {val_accuracy:.2f}%"
    )

training_time = time.time() - start_time
print(f"Training time: {training_time:.2f} seconds")

# Plot accuracy across epochs
plt.plot(range(1, epochs + 1), train_accs, marker="o", label="Training accuracy")
plt.plot(range(1, epochs + 1), val_accs, marker="o", label="Validation accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.title("GoogLeNet on CIFAR-10")
plt.legend()
plt.grid(True)
plt.show()
```
